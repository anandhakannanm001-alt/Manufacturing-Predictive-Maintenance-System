# Feature Engineering

This notebook performs advanced feature engineering on the cleaned dataset to create predictive features for machine failure.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

sys.path.append('../src')

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print('Libraries imported successfully')
```

## 1. Load Cleaned Data

```python
# Load cleaned dataset
df = pd.read_csv('../data/processed/ai4i2020_cleaned.csv')

print(f"Dataset shape: {df.shape}")
print(f"\nColumn names:\n{df.columns.tolist()}")
print(f"\nFirst few rows:")
df.head()
```

## 2. Initialize Feature Engineer

```python
from feature_engineering import FeatureEngineer

# Initialize engineer
engineer = FeatureEngineer()
print('Feature engineer initialized')
```

## 3. Create Temperature Features

```python
# Create temperature-derived features
df = engineer.create_temperature_features(df)

# Display new temperature features
temp_features = [col for col in df.columns if 'temp' in col.lower()]
print(f"Temperature features created: {temp_features}")
df[temp_features].describe()
```

## 4. Create Power Features

```python
# Create power and torque-related features
df = engineer.create_power_features(df)

# Display new power features
power_features = [col for col in df.columns if 'power' in col.lower() or 'torque' in col.lower()]
print(f"Power features created: {power_features}")
df[power_features].describe()
```

## 5. Create Wear Features

```python
# Create tool wear intensity and degradation features
df = engineer.create_wear_features(df)

# Display new wear features
wear_features = [col for col in df.columns if 'wear' in col.lower()]
print(f"Wear features created: {wear_features}")
df[wear_features].describe()
```

## 6. Create Rolling Features

```python
# Create rolling window statistics
df = engineer.create_rolling_features(df, window=5)

# Display new rolling features
rolling_features = [col for col in df.columns if 'rolling' in col.lower()]
print(f"Rolling features created: {len(rolling_features)}")
print(f"Sample rolling features: {rolling_features[:5]}")
```

## 7. Create Health Index

```python
# Create composite health index
df = engineer.create_health_index(df)

# Display health index statistics
print("Health Index Statistics:")
print(df['health_index'].describe())

# Plot health index distribution
plt.figure(figsize=(10, 5))
sns.histplot(df['health_index'], kde=True, bins=30)
plt.title('Health Index Distribution')
plt.xlabel('Health Index (0=Good, 1=Critical)')
plt.show()
```

## 8. Create Anomaly Features

```python
# Create anomaly detection features
df = engineer.create_anomaly_features(df)

# Display new anomaly features
anomaly_features = [col for col in df.columns if 'anomaly' in col.lower() or 'stability' in col.lower() or 'variability' in col.lower()]
print(f"Anomaly features created: {anomaly_features}")
```

## 9. Create Interaction Features

```python
# Create interaction features between key variables
df = engineer.create_interaction_features(df)

# Display new interaction features
interaction_features = [col for col in df.columns if 'interaction' in col.lower()]
print(f"Interaction features created: {interaction_features}")
```

## 10. Run Full Feature Engineering Pipeline

```python
# Apply full pipeline (includes all features)
df_engineered = engineer.full_pipeline(df)

print(f"Final dataset shape: {df_engineered.shape}")
print(f"Total features: {df_engineered.shape[1]}")
```

## 11. Feature Summary

```python
print("\nFeature Summary:")
print(df_engineered.info())

print("\nNew Features Created:")
original_cols = ['udi', 'product_id', 'type', 'air_temperature_k', 'process_temperature_k', 
                'rotational_speed_rpm', 'torque_nm', 'tool_wear_min', 'machine_failure',
                'twf', 'hdf', 'pwf', 'osf', 'rnf']
new_features = [col for col in df_engineered.columns if col not in original_cols]
print(f"Total new features: {len(new_features)}")
print(f"\nNew features:\n{new_features}")
```

## 12. Save Engineered Data

```python
# Save engineered features
df_engineered.to_csv('../data/processed/ai4i2020_engineered.csv', index=False)
print("Engineered data saved to ../data/processed/ai4i2020_engineered.csv")
```
